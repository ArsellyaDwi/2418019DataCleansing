# 7. Data Cleaning

## 7.1 Membersihkan Data Teks

Membersihkan dan menyeragamkan data pada kolom teks dengan menghapus spasi berlebih serta menyesuaikan format penulisan nama mahasiswa.

```python
# Membersihkan dan menyeragamkan data pada kolom teks
kolom_teks = [

    # Menentukan kolom teks yang akan dibersihkan
    'ID Praktikum',
    'NIM',
    'Nama Mahasiswa',
    'Mata Praktikum',
    'Modul',
    'Kehadiran',
    'Status Praktikum',
    'Status Kelulusan'
]

# Melakukan proses pembersihan pada setiap kolom teks
for kolom in kolom_teks:

    # Mengubah data menjadi tipe string dan menghapus spasi berlebih
    df[kolom] = (
        df[kolom]
        .astype('string')
        .str.strip()
    )

# Membersihkan format penulisan nama mahasiswa
df['Nama Mahasiswa'] = (
    # Menghapus spasi berlebih di antara kata
    df['Nama Mahasiswa']
    .str.replace(
        r'\s+',
        ' ',
        regex=True
    )
    # Menyeragamkan huruf awal setiap kata menjadi huruf kapital
    .str.title()
)

# Menampilkan informasi bahwa proses cleaning data teks telah selesai
print("Data teks berhasil dibersihkan.")
```

## 7.2 Membersihkan Mata Praktikum

Menyeragamkan penulisan nama mata praktikum agar memiliki format yang konsisten.

```python
# Membuat fungsi untuk menormalisasi penulisan Mata Praktikum
def normalisasi_kunci(value):

    # Mengecek apakah nilai kosong
    if pd.isna(value):
        return pd.NA

    # Menghapus spasi, mengubah huruf menjadi lowercase,
    # dan menghilangkan spasi berlebih pada data
    return re.sub(
        r'\s+',
        '',
        str(value).strip().lower()
    )


# Membuat mapping untuk menyeragamkan nama Mata Praktikum
mapping_mata = {
    # Menyeragamkan penulisan Algoritma dan Pemrograman
    'algoritmadanpemrograman':
        'Algoritma dan Pemrograman',

    # Menyeragamkan penulisan Sistem Multimedia
    'sistemmultimedia':
        'Sistem Multimedia',

    # Menyeragamkan penulisan Object Oriented Programming
    'objectorientedprogramming':
        'Object Oriented Programming',

    # Menyeragamkan penulisan Basis Data
    'basisdata':
        'Basis Data',

    # Menyeragamkan penulisan Pemrograman Web
    'pemrogramanweb':
        'Pemrograman Web',

    # Memperbaiki variasi penulisan Pemrograman Web
    'pemogramanweb':
        'Pemrograman Web',

    # Menyeragamkan penulisan Jaringan Komputer
    'jaringankomputer':
        'Jaringan Komputer',

    # Menyeragamkan penulisan Kecerdasan Buatan
    'kecerdasanbuatan':
        'Kecerdasan Buatan',

    # Menyeragamkan penulisan Animasi
    'animasi':
        'Animasi',

    # Menyeragamkan penulisan Pengolahan Citra Digital
    'pengolahancitradigital':
        'Pengolahan Citra Digital',

    # Menyeragamkan penulisan Sistem Embedded dan Mikrokontroler
    'sistemembeddeddanmikrokontroler':
        'Sistem Embedded dan Mikrokontroler',

    # Menyeragamkan penulisan Game Developer
    'gamedeveloper':
        'Game Developer',

    # Menyeragamkan penulisan Mobile Programming
    'mobileprogramming':
        'Mobile Programming'
}


# Menerapkan mapping untuk menyeragamkan data Mata Praktikum
df['Mata Praktikum'] = (
    df['Mata Praktikum']
    .apply(
        # Mengubah data berdasarkan mapping yang telah dibuat
        # Jika tidak ditemukan, nilai asli tetap digunakan
        lambda x: mapping_mata.get(
            normalisasi_kunci(x),
            x
        )
    )
)

# Menampilkan informasi bahwa proses standardisasi telah selesai
print("Mata Praktikum berhasil distandarkan.")
```

```python
# Menampilkan daftar Mata Praktikum setelah proses cleaning
print("Daftar Mata Praktikum setelah cleaning:")

print(
    sorted(
        df['Mata Praktikum']
        .dropna()
        .unique()
    )
)
```

## 7.3 Membersihkan Modul

Menyeragamkan format penulisan data Modul dan menangani data yang kosong.

```python
# Membersihkan dan mengisi data kosong pada kolom Modul
df['Modul'] = (

    df['Modul']
    # Mengubah data menjadi tipe string
    .astype('string')
    # Menghapus spasi di awal dan akhir data
    .str.strip()
    # Menyeragamkan huruf awal setiap kata menjadi kapital
    .str.title()

)

# Mengisi data Modul yang kosong dengan keterangan "Tidak Diisi"
df['Modul'] = (
    df['Modul']
    .fillna('Tidak Diisi')

)

# Menampilkan informasi bahwa proses cleaning Modul telah selesai
print("Modul berhasil dibersihkan.")

# Mengecek jumlah data kosong setelah proses cleaning
print(
    "Jumlah data kosong:",
    df['Modul'].isnull().sum()
)
```

## 7.4 Membersihkan Tanggal Praktikum

Menyeragamkan berbagai format tanggal agar menjadi format yang konsisten serta menangani data tanggal yang kosong atau tidak valid.

```python
# Membuat fungsi untuk membersihkan dan menyeragamkan data tanggal
def bersihkan_tanggal(value):

    # Mengecek apakah nilai tanggal kosong
    if pd.isna(value):
        return pd.NaT

    # Mempertahankan nilai yang sudah berupa Timestamp
    if isinstance(value, pd.Timestamp):
        return value

    # Mengubah nilai menjadi string dan menghapus spasi
    value = str(value).strip()

    # Menentukan berbagai format tanggal yang digunakan dalam dataset
    format_tanggal = [
        '%d/%m/%Y',
        '%d-%m-%Y',
        '%Y/%m/%d',
        '%Y-%m-%d',
        '%d %B %Y',
        '%d %b %Y'
    ]

    # Mencoba setiap format tanggal yang tersedia
    for format_date in format_tanggal:

        # Mencoba mengubah nilai menjadi format tanggal
        try:
            return pd.to_datetime(
                value,
                format=format_date
            )

        # Jika format tidak sesuai, mencoba format berikutnya
        except (ValueError, TypeError):
            continue

    # Menandai data yang tidak dapat dikonversi sebagai tanggal
    return pd.NaT
```

```python
# Menerapkan fungsi pembersihan tanggal pada kolom Tanggal Praktikum
tanggal_hasil = (
    df['Tanggal Praktikum']
    .apply(bersihkan_tanggal)
)

# Menghitung jumlah tanggal yang gagal dikonversi
print(
    "Jumlah tanggal gagal diparse:",
    tanggal_hasil.isna().sum()
)
```

```python
# Mengubah hasil tanggal menjadi format DD-MM-YYYY
df['Tanggal Praktikum'] = (
    tanggal_hasil
    .dt.strftime('%d-%m-%Y')
)

# Menampilkan informasi bahwa tanggal berhasil distandarkan
print("Tanggal berhasil distandarkan.")
```

## 7.5 Membersihkan Nilai

Membersihkan format data Nilai, mengubahnya menjadi numerik, memvalidasi rentang nilai 0–100, serta menangani data kosong atau tidak valid.

```python
# Membersihkan dan mengonversi data Nilai menjadi tipe numerik
df['Nilai'] = (
    df['Nilai']
    # Mengubah data menjadi tipe string
    .astype('string')
    # Mengganti tanda koma menjadi titik pada nilai desimal
    .str.replace(
        ',',
        '.',
        regex=False
    )
)

# Mengubah data Nilai menjadi tipe numerik
df['Nilai'] = pd.to_numeric(
    df['Nilai'],
    # Mengubah data yang tidak valid menjadi NaN
    errors='coerce'
)

# Menandai nilai di luar rentang 0-100 sebagai data tidak valid
df.loc[
    (df['Nilai'] < 0) |
    (df['Nilai'] > 100),
    'Nilai'
] = np.nan

# Menghitung nilai median sebagai pengganti data yang kosong atau tidak valid
median_nilai = df['Nilai'].median()

# Mengisi data Nilai yang kosong dengan nilai median
df['Nilai'] = (
    df['Nilai']
    .fillna(median_nilai)
)

# Menampilkan nilai median yang digunakan
print(
    "Median nilai:",
    median_nilai
)

# Mengecek jumlah data Nilai yang masih kosong
print(
    "Jumlah nilai kosong:",
    df['Nilai'].isnull().sum()
)
```

## 7.6 Membersihkan Kehadiran

Menyeragamkan data Kehadiran berdasarkan kategori Hadir, Izin, dan Alpha serta menangani data yang kosong.

```python
# Membuat mapping untuk menyeragamkan data Kehadiran
mapping_kehadiran = {

    # Menyeragamkan kode H menjadi Hadir
    'h': 'Hadir',

    # Menyeragamkan penulisan hadir
    'hadir': 'Hadir',

    # Menyeragamkan kode I menjadi Izin
    'i': 'Izin',

    # Menyeragamkan penulisan izin
    'izin': 'Izin',

    # Menyeragamkan kode A menjadi Alpha
    'a': 'Alpha',

    # Menyeragamkan penulisan alpha
    'alpha': 'Alpha'
}

# Membersihkan dan menyeragamkan data pada kolom Kehadiran
df['Kehadiran'] = (
    df['Kehadiran']
    # Mengubah data menjadi huruf kecil
    .astype('string')
    .str.lower()
    # Menghapus spasi di awal dan akhir data
    .str.strip()
    # Mengubah data berdasarkan mapping yang telah dibuat
    .map(mapping_kehadiran)
    # Mengisi data yang kosong atau tidak sesuai dengan keterangan
    .fillna('Tidak Diisi')
)

# Menampilkan informasi bahwa proses cleaning Kehadiran telah selesai
print("Kehadiran berhasil dibersihkan.")

# Mengecek jumlah data kosong setelah proses cleaning
print(
    "Jumlah data kosong:",
    df['Kehadiran'].isnull().sum()
)
```

## 7.7 Membersihkan Durasi Praktikum

Menyeragamkan berbagai format durasi praktikum menjadi satuan menit dan menangani data yang kosong atau tidak valid.

```python
# Membuat fungsi untuk membersihkan dan mengonversi data durasi praktikum
def convert_duration(value):

    # Mengecek apakah data durasi kosong
    if pd.isna(value):
        return np.nan

    # Mengubah data Timedelta menjadi satuan menit
    if isinstance(value, pd.Timedelta):
        return value.total_seconds() / 60

    # Mengubah data menjadi string, huruf kecil, dan menghapus spasi
    value = str(value).strip().lower()

    # Mengecek apakah durasi menggunakan satuan menit
    if 'menit' in value:

        try:
            # Menghapus teks "menit" dan mengambil angka durasi
            angka = (
                value
                .replace('menit', '')
                .strip()
            )

            return float(angka)

        # Jika data tidak dapat dikonversi, mengembalikan NaN
        except:
            return np.nan

    # Mengecek apakah durasi menggunakan satuan jam
    if 'jam' in value:

        try:
            # Menghapus teks "jam" dan mengambil angka durasi
            angka = (
                value
                .replace('jam', '')
                .strip()
            )

            # Mengubah jam menjadi menit
            return float(angka) * 60

        # Jika data tidak dapat dikonversi, mengembalikan NaN
        except:
            return np.nan

    # Mengecek format waktu seperti 01:30
    if re.fullmatch(
        r'\d+:\d+',
        value
    ):

        try:
            # Memisahkan data menjadi jam dan menit
            jam, menit = value.split(':')

            # Mengubah seluruh durasi menjadi menit
            return (
                int(jam) * 60
                + int(menit)
            )

        # Jika format tidak dapat diproses, mengembalikan NaN
        except:
            return np.nan

    # Mencoba mengubah data numerik langsung menjadi angka
    try:
        return float(value)

    # Jika data tidak valid, mengembalikan NaN
    except:
        return np.nan


# Menerapkan fungsi cleaning pada kolom Durasi Praktikum
df['Durasi Praktikum (Menit)'] = (
    df['Durasi Praktikum (Menit)']
    .apply(convert_duration)
)
```

```python
# Menandai durasi yang bernilai 0 atau negatif sebagai data tidak valid
df.loc[
    df['Durasi Praktikum (Menit)'] <= 0,
    'Durasi Praktikum (Menit)'
] = np.nan

# Menghitung nilai median dari durasi yang valid
median_durasi = (
    df['Durasi Praktikum (Menit)']
    .median()
)

# Mengisi data durasi yang kosong dengan nilai median
# Kemudian membulatkan dan mengubahnya menjadi bilangan bulat
df['Durasi Praktikum (Menit)'] = (
    df['Durasi Praktikum (Menit)']
    .fillna(median_durasi)
    .round()
    .astype(int)
)

# Menampilkan nilai median yang digunakan
print(
    "Median durasi:",
    median_durasi
)

# Menampilkan informasi bahwa proses cleaning telah selesai
print("Durasi berhasil dibersihkan.")
```

## 7.8 Membersihkan Jumlah Percobaan

Membersihkan format data Jumlah Percobaan dengan menghapus satuan dan mengubahnya menjadi data numerik.

```python
# Membersihkan dan mengonversi data Jumlah Percobaan
df['Jumlah Percobaan'] = (
    df['Jumlah Percobaan']
    # Mengubah data menjadi string dan huruf kecil
    .astype('string')
    .str.lower()
    # Menghapus teks "kali" dari data
    .str.replace(
        'kali',
        '',
        regex=False
    )
    # Menghapus simbol "x" dari data
    .str.replace(
        'x',
        '',
        regex=False
    )
    # Menghapus spasi di awal dan akhir data
    .str.strip()
)

# Mengubah data Jumlah Percobaan menjadi tipe numerik
df['Jumlah Percobaan'] = pd.to_numeric(
    df['Jumlah Percobaan'],
    # Mengubah data yang tidak valid menjadi NaN
    errors='coerce'
)

# Menampilkan informasi bahwa proses cleaning telah selesai
print(
    "Jumlah Percobaan berhasil dibersihkan."
)
```
